```python
import pandas as pd
from sqlalchemy import create_engine
from dotenv import load_dotenv
import os

load_dotenv()
password = os.getenv('DB_PASSWORD')

engine = create_engine(f'postgresql://postgres:{password}@localhost:5432/olist_ecommerce')

# quick test
print(engine.connect())
```

```python
from sqlalchemy import text

with engine.connect() as conn:
    conn.execute(text("""
        INSERT INTO category_translation (product_category_name, product_category_name_english)
        VALUES ('unknown', 'unknown')
    """))
    conn.commit()

print("Inserted 'unknown' into category_translation")
```

```python
PROCESSED_DIR = '../data/processed'

load_order = [
    ('customers', 'customers_clean.csv'),
    ('sellers', 'sellers_clean.csv'),
    ('geolocation', 'geolocation_clean.csv'),
    ('products', 'products_clean.csv'),
    ('orders', 'orders_clean.csv'),
    ('order_items', 'order_items_clean.csv'),
    ('payments', 'payments_clean.csv'),
    ('reviews', 'reviews_clean.csv'),
]

for table_name, filename in load_order:
    df = pd.read_csv(f'{PROCESSED_DIR}/{filename}')
    df.to_sql(table_name, engine, if_exists='append', index=False)
    print(f'{table_name:15s} loaded -> {len(df)} rows')
```

```python
reviews_df = pd.read_csv('../data/processed/reviews_clean.csv')
reviews_df.to_sql('reviews', engine, if_exists='append', index=False)
print(f'reviews loaded -> {len(reviews_df)} rows')
```
